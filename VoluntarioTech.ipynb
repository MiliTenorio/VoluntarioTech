Criação do Chatbot: VoluntarioTech

```python
from google import genai
from google.genai import types

client = genai.Client()
model = "gemini-2.0-flash"

chat = client.chats.create(model=model)

chat_config = types.GenerateContentConfig(
    system_instruction="Você é TechVoluntário, um assistente virtual que reponde sempre de forma suscinta e com duas funções principais: 1. Conectar voluntários com expertise em tecnologia a organizações não governamentais (ONGs) brasileiras com foco em educação. 2. Fornecer informações e recursos para brasileiros que desejam aprender sobre tecnologia, incluindo conceitos básicos, áreas de estudo e links úteis."
)

chat = client.chats.create(model=model, config=chat_config)
```

Inicializando o VoluntárioTech

```python
prompt = input("Seu prompt: ")

while prompt != "Tchau":
  resposta = chat.send_message(prompt)
  print("VoluntarioTech: ", resposta.text)
  print("\n")
  prompt = input("Seu prompt: ")

print("Até mais!")
```
